# Извлечение признаков FEI Face с предобученным ResNet + Decoder

Подход:
1. **Encoder**: Предобученный ResNet-18 (заморожен) → 512-мерный вектор
2. **Decoder**: Обучаемый свёрточный декодер (128x128 RGB)
3. **Loss**: MSE + Perceptual Loss (VGG) → чёткие изображения без замыливания

Результат: 512-мерные признаки совместимы с NCT pipeline

```python
import os, glob, re
import numpy as np
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
from PIL import Image

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Device: {device}')
```

```python
import kagglehub
path = kagglehub.dataset_download("whizzkid/fei-face-data")
print("Path to dataset files:", path)
```

```python
# --- Константы ---
FEI_ROOT = "/kaggle/input/fei-face-data"
IMG_SIZE = 128
LATENT_DIM = 512
BATCH_SIZE = 64
NUM_EPOCHS = 80
LR = 1e-3
```

```python
# --- Dataset ---
class FEIFaceDataset(Dataset):
    def __init__(self, root, transform=None):
        self.transform = transform
        parts = [os.path.join(root, f"originalimages_part{i}") for i in range(1, 5)]
        self.files = []
        self.labels = []
        for p in parts:
            if os.path.isdir(p):
                for path in sorted(glob.glob(os.path.join(p, "*.jpg"))):
                    fname = os.path.basename(path)
                    match = re.match(r"(\d+)-(\d+)\.jpg", fname)
                    if match:
                        pid = int(match.group(1))
                        if 1 <= pid <= 200:
                            self.files.append(path)
                            self.labels.append(pid - 1)
        print(f"Загружено {len(self.files)} изображений, {len(set(self.labels))} лиц")

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        img = Image.open(self.files[idx]).convert("RGB")
        if self.transform:
            img = self.transform(img)
        return img, self.labels[idx]


transform = transforms.Compose([
    transforms.CenterCrop(480),
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
])

dataset = FEIFaceDataset(root=FEI_ROOT, transform=transform)
loader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=2, pin_memory=True)
```

## Архитектура: ResNet-encoder + Decoder

```python
class ResNetEncoder(nn.Module):
    """Предобученный ResNet-18 как энкодер → 512-мерный вектор"""
    def __init__(self, latent_dim=512, freeze_backbone=True):
        super().__init__()
        resnet = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
        # Всё кроме последнего FC слоя
        self.backbone = nn.Sequential(*list(resnet.children())[:-1])  # output: (B, 512, 1, 1)
        
        if freeze_backbone:
            for param in self.backbone.parameters():
                param.requires_grad = False
        
        self.fc = nn.Linear(512, latent_dim)
    
    def forward(self, x):
        h = self.backbone(x)           # (B, 512, 1, 1)
        h = h.view(h.size(0), -1)      # (B, 512)
        z = self.fc(h)                 # (B, latent_dim)
        return z


class ConvDecoder(nn.Module):
    """Свёрточный декодер: 512 → 128x128 RGB"""
    def __init__(self, latent_dim=512):
        super().__init__()
        self.fc = nn.Linear(latent_dim, 256 * 8 * 8)
        
        self.decoder = nn.Sequential(
            # 8x8 → 16x16
            nn.ConvTranspose2d(256, 256, 4, 2, 1),
            nn.BatchNorm2d(256),
            nn.ReLU(True),
            
            # 16x16 → 32x32
            nn.ConvTranspose2d(256, 128, 4, 2, 1),
            nn.BatchNorm2d(128),
            nn.ReLU(True),
            
            # 32x32 → 64x64
            nn.ConvTranspose2d(128, 64, 4, 2, 1),
            nn.BatchNorm2d(64),
            nn.ReLU(True),
            
            # 64x64 → 128x128
            nn.ConvTranspose2d(64, 32, 4, 2, 1),
            nn.BatchNorm2d(32),
            nn.ReLU(True),
            
            nn.Conv2d(32, 3, 3, 1, 1),
            nn.Sigmoid(),
        )
    
    def forward(self, z):
        h = self.fc(z)
        h = h.view(-1, 256, 8, 8)
        return self.decoder(h)


class ResNetAutoencoder(nn.Module):
    def __init__(self, latent_dim=512, freeze_backbone=True):
        super().__init__()
        self.encoder = ResNetEncoder(latent_dim, freeze_backbone)
        self.decoder = ConvDecoder(latent_dim)
    
    def forward(self, x):
        z = self.encoder(x)
        recon = self.decoder(z)
        return recon, z
    
    def encode(self, x):
        return self.encoder(x)
    
    def decode(self, z):
        return self.decoder(z)
```

## Perceptual Loss (VGG-based)

```python
class PerceptualLoss(nn.Module):
    """Perceptual loss на основе VGG-16 для чётких реконструкций"""
    def __init__(self):
        super().__init__()
        vgg = models.vgg16(weights=models.VGG16_Weights.DEFAULT).features
        
        # Берём первые 3 блока VGG для разных масштабов
        self.block1 = nn.Sequential(*list(vgg)[:4]).eval()   # relu1_2
        self.block2 = nn.Sequential(*list(vgg)[:9]).eval()   # relu2_2
        self.block3 = nn.Sequential(*list(vgg)[:16]).eval()  # relu3_3
        
        for param in self.parameters():
            param.requires_grad = False
        
        self.register_buffer('mean', torch.tensor([0.485, 0.456, 0.406]).view(1, 3, 1, 1))
        self.register_buffer('std', torch.tensor([0.229, 0.224, 0.225]).view(1, 3, 1, 1))
    
    def normalize(self, x):
        return (x - self.mean) / self.std
    
    def forward(self, recon, target):
        recon_n = self.normalize(recon)
        target_n = self.normalize(target)
        
        loss = 0
        loss += F.mse_loss(self.block1(recon_n), self.block1(target_n))
        loss += F.mse_loss(self.block2(recon_n), self.block2(target_n))
        loss += F.mse_loss(self.block3(recon_n), self.block3(target_n))
        return loss
```

## Обучение

```python
model = ResNetAutoencoder(latent_dim=LATENT_DIM, freeze_backbone=True).to(device)
perceptual_loss_fn = PerceptualLoss().to(device)

# Обучаем только fc-слой энкодера + весь декодер
trainable_params = list(model.encoder.fc.parameters()) + list(model.decoder.parameters())
optimizer = optim.Adam(trainable_params, lr=LR)
scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=30, gamma=0.5)

total_params = sum(p.numel() for p in trainable_params)
print(f'Обучаемых параметров: {total_params:,}')
```

```python
LAMBDA_PERCEPTUAL = 0.5

model.train()
for epoch in range(NUM_EPOCHS):
    epoch_loss = 0
    epoch_mse = 0
    epoch_perc = 0
    
    for imgs, _ in loader:
        imgs = imgs.to(device)
        optimizer.zero_grad()
        
        recon, z = model(imgs)
        
        mse = F.mse_loss(recon, imgs)
        perc = perceptual_loss_fn(recon, imgs)
        loss = mse + LAMBDA_PERCEPTUAL * perc
        
        loss.backward()
        optimizer.step()
        
        epoch_loss += loss.item()
        epoch_mse += mse.item()
        epoch_perc += perc.item()
    
    scheduler.step()
    n_batches = len(loader)
    
    if epoch % 10 == 0 or epoch == NUM_EPOCHS - 1:
        print(f'Epoch [{epoch+1}/{NUM_EPOCHS}] '
              f'Loss: {epoch_loss/n_batches:.4f} '
              f'(MSE: {epoch_mse/n_batches:.4f}, '
              f'Perceptual: {epoch_perc/n_batches:.4f})')

torch.save({
    'model_state_dict': model.state_dict(),
    'latent_dim': LATENT_DIM,
}, 'checkpoint_resnet_ae.pth')
print('Модель сохранена: checkpoint_resnet_ae.pth')
```

## Визуализация реконструкций

```python
model.eval()
with torch.no_grad():
    imgs, labels = next(iter(loader))
    imgs = imgs[:8].to(device)
    recon, z = model(imgs)

fig, axes = plt.subplots(2, 8, figsize=(20, 5))
for i in range(8):
    axes[0, i].imshow(imgs[i].cpu().permute(1, 2, 0))
    axes[0, i].set_title('Original')
    axes[0, i].axis('off')
    
    axes[1, i].imshow(recon[i].cpu().permute(1, 2, 0).clamp(0, 1))
    axes[1, i].set_title('Reconstructed')
    axes[1, i].axis('off')

plt.suptitle('ResNet Encoder + Decoder (Perceptual Loss)', fontsize=14)
plt.tight_layout()
plt.show()
```

## Дообучение: разморозка части ResNet (fine-tuning)

```python
# Размораживаем последние 2 блока ResNet для fine-tuning
for name, param in model.encoder.backbone.named_parameters():
    if 'layer3' in name or 'layer4' in name:
        param.requires_grad = True

optimizer_ft = optim.Adam(filter(lambda p: p.requires_grad, model.parameters()), lr=1e-4)
scheduler_ft = optim.lr_scheduler.StepLR(optimizer_ft, step_size=15, gamma=0.5)

ft_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f'Fine-tuning параметров: {ft_params:,}')

model.train()
for epoch in range(40):
    epoch_loss = 0
    for imgs, _ in loader:
        imgs = imgs.to(device)
        optimizer_ft.zero_grad()
        
        recon, z = model(imgs)
        mse = F.mse_loss(recon, imgs)
        perc = perceptual_loss_fn(recon, imgs)
        loss = mse + LAMBDA_PERCEPTUAL * perc
        
        loss.backward()
        optimizer_ft.step()
        epoch_loss += loss.item()
    
    scheduler_ft.step()
    if epoch % 10 == 0 or epoch == 39:
        print(f'FT Epoch [{epoch+1}/40] Loss: {epoch_loss/len(loader):.4f}')

torch.save({
    'model_state_dict': model.state_dict(),
    'latent_dim': LATENT_DIM,
}, 'checkpoint_resnet_ae_finetuned.pth')
print('Fine-tuned модель сохранена')
```

```python
# Визуализация после fine-tuning
model.eval()
with torch.no_grad():
    imgs, labels = next(iter(loader))
    imgs = imgs[:8].to(device)
    recon, z = model(imgs)

fig, axes = plt.subplots(2, 8, figsize=(20, 5))
for i in range(8):
    axes[0, i].imshow(imgs[i].cpu().permute(1, 2, 0))
    axes[0, i].set_title('Original')
    axes[0, i].axis('off')
    
    axes[1, i].imshow(recon[i].cpu().permute(1, 2, 0).clamp(0, 1))
    axes[1, i].set_title('Reconstructed')
    axes[1, i].axis('off')

plt.suptitle('После Fine-tuning (ResNet + Perceptual Loss)', fontsize=14)
plt.tight_layout()
plt.show()
```

## Извлечение признаков для NCT pipeline

```python
# Извлечение упорядоченных признаков (200 классов x 14 изображений x 512 признаков)
model.eval()
all_loader = DataLoader(dataset, batch_size=256, shuffle=False, num_workers=2)

latents = []
labels_all = []
with torch.no_grad():
    for imgs, lbls in all_loader:
        imgs = imgs.to(device)
        z = model.encode(imgs)
        latents.append(z.cpu().numpy())
        labels_all.append(lbls.numpy())

X_latent = np.vstack(latents)
y_labels = np.hstack(labels_all)
print(f'Латентное пространство: {X_latent.shape}, Метки: {y_labels.shape}')
```

```python
# Упорядочивание по классам
X_ordered = []
for class_id in range(200):
    indices = np.where(y_labels == class_id)[0]
    X_class = X_latent[indices]
    assert X_class.shape == (14, 512), f"Класс {class_id}: {X_class.shape}"
    X_ordered.append(X_class)

X_3d = np.stack(X_ordered)  # (200, 14, 512)
print(f'Итоговый массив: {X_3d.shape}')

np.save('resnet_ae_classes_3d_fei_512.npy', X_3d)

# CSV для NCT (плоский формат)
X_2d = X_3d.reshape(-1, 512)  # (2800, 512)
np.savetxt('resnet_ae_3d_data.csv', X_2d, delimiter=',')
print(f'CSV: {X_2d.shape}')

from IPython.display import FileLink
display(FileLink('resnet_ae_classes_3d_fei_512.npy'))
display(FileLink('resnet_ae_3d_data.csv'))
display(FileLink('checkpoint_resnet_ae_finetuned.pth'))
```

## Сравнение с VAE

```python
# Метрики качества реконструкции
model.eval()
total_mse = 0
total_psnr = 0
n = 0

with torch.no_grad():
    for imgs, _ in all_loader:
        imgs = imgs.to(device)
        recon, _ = model(imgs)
        mse = F.mse_loss(recon, imgs, reduction='none').mean(dim=[1,2,3])
        psnr = -10 * torch.log10(mse + 1e-8)
        total_mse += mse.sum().item()
        total_psnr += psnr.sum().item()
        n += imgs.size(0)

print(f'Средний MSE: {total_mse/n:.6f}')
print(f'Средний PSNR: {total_psnr/n:.2f} dB')
print(f'(Для справки: PSNR > 25 dB — хорошо, > 30 dB — отлично)')
```

```python
# t-SNE визуализация латентного пространства
from sklearn.manifold import TSNE
import seaborn as sns

n_sample = min(2000, len(X_latent))
idx = np.random.choice(len(X_latent), n_sample, replace=False)

tsne = TSNE(n_components=2, perplexity=30, n_iter=500, random_state=42)
Z_tsne = tsne.fit_transform(X_latent[idx])

plt.figure(figsize=(10, 8))
sns.scatterplot(x=Z_tsne[:, 0], y=Z_tsne[:, 1], hue=y_labels[idx], palette='tab20', s=50, legend=False)
plt.title('t-SNE: ResNet Autoencoder Latent Space (FEI Face)')
plt.axis('off')
plt.show()
```
